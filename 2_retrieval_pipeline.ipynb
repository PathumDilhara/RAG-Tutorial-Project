# **2) Retreival pipeline**

```python
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_ollama import ChatOllama
```

### **Configurations**

```python
db_path = "db/chroma_db"
```

### **Load same embedding model**

```python
# loading a sentence embedding model:

embedding_model = HuggingFaceEmbeddings(
    model_name="saved_models/all-mpnet-base-v2", # all-MiniLM-L12-v2, all-MiniLM-L6-v2, all-mpnet-base-v2
    model_kwargs={"device":"cpu"},
    encode_kwargs={"normalize_embeddings":True}
)
```

### **Connect Vector Database**

```python
# loading + connecting to the existing database that created in ingestion

db = Chroma(
    persist_directory=db_path,
    embedding_function=embedding_model,
    collection_metadata={"hnsw:space":"cosine"}
)
```

```python
retreiver = db.as_retriever(search_kwargs={"k":5}) # 5 matching chunks will get
```

### **Getting relevant docs chunks as text**

```python
query = "In what year tesla begin production of the roadster?"
relevant_docs = retreiver.invoke(query)
```

```python
for i, doc in enumerate(relevant_docs, 1):
    print(f"### Document {i} : \n {doc.page_content}\n------------------\n")
```

### **LLM**

```python
llm = ChatOllama(model="llama3")
```
